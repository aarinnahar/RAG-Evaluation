```python
def SortedTargetMatch(arr, target):
    # 1. Create your two pointers (indexes)
    left = 0
    right = len(arr) - 1
    
    # 2. Loop until the pointers meet
    while left < right:
        current_sum = arr[left] + arr[right]
        
        # 3. Write the 3 conditions (Match, Too Small, Too Big)
        if current_sum == target:
            return "true"
        elif current_sum < target:
            # How do you move the left pointer?
            left = left + 1
        else:
            # How do you move the right pointer?
            right = right - 1
            
    # If the loop finishes and they cross, no pair was found
    return "false"
```

```python
# DRILL 1

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score



def e2e_pipeline(df, target):
    label = df.iloc[0:, -1]
    training = df.iloc[:,:-1]
    X_Train, x_test, y_train, y_test = train_test_split(training, label, test_size=0.2, random_state=42)
    z = StandardScaler()
    scaled_x_train = z.fit_transform(X_Train)
    scaled_x_test = z.transform(x_test)
    model = RandomForestClassifier(n_estimators = 50, 
                                   criterion='gini',
                                   max_depth = 3)
    model.fit(scaled_x_train,y_train)
    y_pred = model.predict(scaled_x_test)
    accuracy_scores = round((accuracy_score(y_test, y_pred)), 4)

    return accuracy_scores
```
